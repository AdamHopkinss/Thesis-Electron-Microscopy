# Thesis structure notes

## High-level structure

The thesis should follow a reader-oriented structure, not the exact order in which I worked.

The main flow should be:

1. Problem
2. Background theory needed to understand the problem
3. Method used in this thesis
4. Results from the method
5. Discussion and interpretation
6. Conclusions and future work

A good overall chapter structure is:

1. Introduction
2. Background
3. Methodology
4. Results / Numerical Experiments
5. Discussion
6. Conclusions and Future Work

The abstract and Swedish summary are written last, even if they appear first in the thesis.

---

## Core principle

The thesis should feel like:

CT is difficult and noise-sensitive  
-> standard reconstruction methods involve a trade-off  
-> SIAC may help  
-> here is the theory behind SIAC  
-> here is how I apply it  
-> here are the results  
-> here is what the results mean

So the structure should reflect:

* what the problem is
* what theory is needed
* what I actually did
* what happened
* what I learned

---

## What goes where

### 1. Introduction

Purpose: motivate the thesis and explain the research question.

Include:

* a short introduction to CT reconstruction
* why reconstruction is difficult
* the trade-off between sharpness and noise suppression
* why SIAC is interesting in this setting
* a short description of my contribution
* outline of the thesis

Do **not** overload this chapter with detailed theory or equations.

---

### 2. Background

Purpose: provide the theory needed for the rest of the thesis.

This chapter should contain the mathematics and concepts needed to understand the method and results.

Suggested sections:

#### 2.1 Computed Tomography and FBP

Include:

* Radon transform
* Fourier slice theorem if needed
* filtered backprojection
* Ram--Lak filter
* windowed filters such as Hann

This section should explain the standard CT reconstruction framework that I compare against.

#### 2.2 Ill-posedness in CT

Keep this short.

Include:

* why inversion is sensitive to noise
* why high frequencies are problematic
* why filtering matters

Since I do not actually perform regularization methods such as Tikhonov or TV, I should stop before a full regularization theory section.

#### 2.3 SIAC filtering

This is the main theoretical background for the thesis.

Suggested subsections:

* historical background
* polynomial reproduction
* moment conditions
* B-splines and SIAC kernel construction
* properties of SIAC filters
* why SIAC may be useful in CT

Important idea:
SIAC is not just smoothing. It is smoothing with polynomial reproduction, so it may preserve useful structure better than standard low-pass filtering.

#### 2.4 Data representations

Optional but relevant.

Include:

* nodal representation
* modal representation
* nodal-to-modal interpretation
* why the representation matters for DG--SIAC

This should remain conceptual and not become an implementation dump.

---

### 3. Methodology

Purpose: explain exactly what I do in this thesis.

This chapter should contain the practical setup and reconstruction strategies.

Suggested sections:

#### 3.1 Problem setup

Include:

* Shepp--Logan phantom
* geometry
* noise model
* reconstruction domain
* detector and angle setup

#### 3.2 SIAC application strategies

This is where I explain the actual approaches used in the thesis.

Include:

* DG--SIAC postprocessing
* Fourier--SIAC postprocessing
* detector-space SIAC or SIAC inside the reconstruction filter

This chapter should explain how each method is applied.

#### 3.3 Parameter choices

Include:

* reproduction order r
* B-spline order n
* DG degree p when relevant
* how parameters are varied in experiments

This is the correct place to state parameter choices clearly.

#### 3.4 Evaluation metrics

Include:

* masked relative L2 error
* gradient-weighted SSIM

Optionally also report:

* standard relative L2
* standard SSIM

Even if the masked metrics are the main ones, including standard metrics can make the work easier to compare with other reconstruction papers.

#### 3.5 Verification

Yes, include verification, but keep it short and focused.

Use it to show:

* the SIAC implementation behaves correctly
* the method works on analytical or smooth test functions before moving to phantom experiments

This improves trust in the numerical experiments.

---

### 4. Results / Numerical Experiments

Purpose: present what happened.

This chapter should show results, not carry too much theory.

Suggested sections:

* baseline reconstructions with Ram--Lak and Hann
* SIAC reconstructions
* parameter studies
* Monte Carlo studies
* performance under different noise levels
* visual comparisons
* robustness analysis

Keep interpretation lighter here.
The chapter should mostly answer:

* what do the reconstructions look like
* how do the metrics behave
* how do parameters affect the result

---

### 5. Discussion

Purpose: explain and interpret the results.

This is where I discuss:

* why SIAC works or does not work
* how it compares to Hann and Ram--Lak
* what trade-offs appear
* what the parameter studies mean
* limitations of the methods
* what is specific to my implementation

Important:
Long explanations that are currently buried inside the experiments chapter may belong here instead.

---

### 6. Conclusions and Future Work

Purpose: close the thesis clearly.

#### Conclusions

Summarize:

* the main research question
* the main findings
* what was shown by the experiments
* what can reasonably be concluded

#### Future work

Include:

* DG-discretized forward operator
* more rigorous analysis of SIAC in CT
* better parameter selection strategies
* real CT data
* alternative data representations
* extensions beyond the current proof-of-concept setting

This is the right place for ideas that were not actually carried out.

---

## What should not go in the wrong chapter

### Do not put these in Background

* implementation details
* actual experiment results
* long parameter sweeps
* Monte Carlo findings

### Do not put these in Results

* large blocks of theory
* derivations that should have been explained earlier
* major conceptual motivation

### Do not put these in Future Work if they are already used

* any method that is actually part of the thesis contribution

---

## About DG forward operator

The DG-discretized forward operator is interesting, but if I did not actually implement and use it as part of the core thesis results, it should remain in Future Work or perhaps be mentioned briefly in Background/Discussion as motivation.

It should not become a central chapter if it is not part of the actual completed work.

---

## About ill-posedness

I should include an ill-posedness section, but keep it focused.

I only need enough to explain:

* why CT inversion is noise-sensitive
* why filters are introduced
* why there is a trade-off between sharpness and smoothing

Since I do not develop or compare explicit regularization methods, I should stop before long regularization theory.

---

## About verification

I should include verification, but briefly.

Good uses:

* SIAC on smooth functions
* polynomial reproduction checks
* simple demonstrations that the implementation is correct

This belongs best in Methodology or as an initial section before the main numerical experiments.

---

## Writing order

Even though the thesis appears in one order, I should probably write in this order:

1. Methodology
2. Results / Numerical Experiments
3. Discussion
4. Background
5. Introduction
6. Abstract
7. Sammanfattning

This makes writing easier because I first write what I actually did.

---

## Final structural rule

Each chapter should answer one main question:

* Introduction: What is the problem and why does it matter?
* Background: What does the reader need to know first?
* Methodology: What exactly did I do?
* Results: What happened?
* Discussion: What does it mean?
* Conclusions/Future Work: What can I claim, and what comes next?
